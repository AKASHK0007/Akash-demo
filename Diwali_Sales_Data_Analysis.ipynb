```python
import pandas as pd

df = pd.read_csv('Diwali Sales Data.csv', encoding='unicode_escape')
display(df.head())
display(df.info())
display(df.describe())
```

```python
df.drop(['Status', 'unnamed1'], axis=1, inplace=True)
df.dropna(subset=['Amount'], inplace=True)
display(df.isnull().sum())
```

```python
sales_age = df.groupby('Age Group')['Amount'].sum().sort_values(ascending=False)
sales_gender = df.groupby('Gender')['Amount'].sum()
sales_marital_status = df.groupby('Marital_Status')['Amount'].sum()
sales_occupation = df.groupby('Occupation')['Amount'].sum().sort_values(ascending=False)
sales_state = df.groupby('State')['Amount'].sum().sort_values(ascending=False)

display("Sales by Age Group:")
display(sales_age)
display("Sales by Gender:")
display(sales_gender)
display("Sales by Marital Status:")
display(sales_marital_status)
display("Sales by Occupation:")
display(sales_occupation)
display("Sales by State:")
display(sales_state)
```

```python
sales_product_category = df.groupby('Product_Category')['Amount'].sum().sort_values(ascending=False)
display(sales_product_category)
```

## Summary:

### Data Analysis Key Findings

*   The "26-35" age group contributed the most to sales.
*   Female customers generated significantly higher sales than males.
*   Customers with '0' marital status (unmarried) contributed more to sales than those with '1' (married).
*   The "IT Sector" occupation group showed the highest sales, followed by "Healthcare" and "Aviation".
*   Uttar Pradesh emerged as the state with the highest sales, followed by Maharashtra and Karnataka.
*   The top-selling product categories were Food, Clothing & Apparel, and Electronics & Gadgets.

### Insights or Next Steps

*   Target marketing efforts towards the 26-35 age group, females, and unmarried individuals, particularly those in the IT, Healthcare, and Aviation sectors.
*   Focus marketing campaigns and product stocking on the top-performing states: Uttar Pradesh, Maharashtra, and Karnataka, and the best-selling product categories: Food, Clothing & Apparel, and Electronics & Gadgets.
